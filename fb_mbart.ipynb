```python
!pip install -q transformers sentencepiece

from transformers import MBartForConditionalGeneration, MBart50TokenizerFast

model_name = "facebook/mbart-large-50-many-to-many-mmt"
tokenizer = MBart50TokenizerFast.from_pretrained(model_name)
model = MBartForConditionalGeneration.from_pretrained(model_name)

hindi_sentence = "भोपाल भारत का एक राज्य है।"

tokenizer.src_lang = "hi_IN"
inputs = tokenizer(hindi_sentence, return_tensors="pt")
translated_tokens = model.generate(**inputs, forced_bos_token_id=tokenizer.lang_code_to_id["te_IN"])
telugu_translation = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)
print("Telugu:", telugu_translation)

tokenizer.src_lang = "te_IN"
inputs = tokenizer(telugu_translation, return_tensors="pt")
translated_tokens = model.generate(**inputs, forced_bos_token_id=tokenizer.lang_code_to_id["hi_IN"])
hindi_back = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)
print("Hindi back:", hindi_back)
```
